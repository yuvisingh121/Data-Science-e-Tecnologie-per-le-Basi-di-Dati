## Progettazione di un data warehouse
Progettare un data warehouse per analizzare le informazioni relative ai volontari che hanno prestato servizio presso le diverse gare disputate durante le Olimpiadi 2026.

### Descrizione del problema
Per ciascun volontario sono disponibili alcune informazioni anagrafiche. In particolare è nota la provenienza, espressa in termini di città, provincia, regione e nazione, e il mese di nascita, con anche l’indicazione del bimestre, trimestre e anno corrispondenti. Si desidera anche memorizzare se per il volontario è la prima partecipazione ad una olimpiade invernale oppure se ha già partecipato ad altre edizioni. 

Si vogliono memorizzare le località presso cui sono svolte le gare. Ciascuna località è descritta in termini di città, provincia e regione corrispondenti. Per ogni città è noto l’elenco delle discipline per cui si svolgono le gare presso la città (ad esempio discesa libera, gigante, superG, short track, ecc.)

Si conosce infine l’elenco delle date per cui si è svolta almeno una gara, insieme con l’indicazione del mese. Sono inoltre noti il giorno della settimana e il mese dell’anno corrispondenti. Si vuole memorizzare anche il tipo di gara svolta, ossia se si tratti di una gara di qualificazione, semifinale o finale.

Si vuole analizzare il numero medio di gare per volontario e l’importo medio di rimborso spese ricevuto per volontario in base alle seguenti informazioni:

- mese, bimestre, trimestre, anno di nascita del volontario 
- città, provincia, regione e nazione di provenienza del volontario
- indicazione se è la prima partecipazione come volontario a un'olimpiade invernale (un valore tra ‘sì’ e ‘no’)
- città, provincia e regione presso cui si svolgono le gare. Per ogni città è noto l’elenco delle discipline per cui si svolgono le gare presso la città (ad esempio discesa libera, gigante, superG, short track, ecc.)
- data e mese in cui si svolgono le gare. Sono inoltre noti il giorno della settimana e il mese dell’anno corrispondenti.
- tipo di gara (un valore tra qualificazione, semifinale, finale).
